```python
df["FamilySize"] = df["SibSp"] + df["Parch"] + 1
```

```python
def family_category(size):
    if size == 1:
        return "Alone"
    elif size <= 4:
        return "Small"
    else:
        return "Large"
```

```python
df["FamilyCategory"] = df["FamilySize"].apply(family_category)
```

```python
def get_title(name):
    parts = name.split(",")
    return parts[1].split(" ")[1].rstrip(".")
```

```python
get_title("Braund, Mr. Owen Harris")
```

```python
df["Title"] = df["Name"].apply(get_title)
```

```python
df["Title"].value_counts()
```

```python
df.groupby("Title")["Survived"].mean()
```
